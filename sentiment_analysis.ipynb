```python
import numpy as np
import pandas as pd
```

```python
df =pd.read_csv("amazonreviews.tsv",sep ="\t")
```

```python
df.head()
```

```python
df['label'].value_counts()
```

```python
df.dropna(inplace=True)
blanks=[]
for i,lb,rv in df.itertuples():
  if type(rv)==str:
    if rv.isspace():
      blanks.append(i)
print(len(blanks),'blanks: ',blanks)
```

```python
sid.polarity_scores(df.loc[0]['review'])
```

```python
df['score']=df['review'].apply(lambda review:sid.polarity_scores(review))
```

```python
df.head()
```

```python
df['compound'] = df['score'].apply(lambda score_dict:score_dict['compound'])
```

```python
df.head()
```

```python
df['comp_score'] = df['compound'].apply(
    lambda c: '*****' if c >= 0.8 else ('***' if c >= 0.5 else '*')
)
```

```python
df['comp_score1'] = df['compound'].apply(lambda c: 'pos' if c>=0 else 'neg')
```

```python
df.head(50)
```

```python
def star(compound):
  if compound>=0.8:
    return '*****'
  elif compound >=0.5:
    return '***'
  else:
    return '*'
```

```python
from sklearn.metrics import classification_report, accuracy_score
```

```python
accuracy_score(df['label'],df['comp_score1'])
```

```python
print(classification_report(df['label'],df['comp_score1']))
```
